```python
student = {}

student["Name"] = input("Enter Name: ")
student["Roll No"] = input("Enter Roll No: ")

student["Math"] = int(input("Math Marks: "))
student["English"] = int(input("English Marks: "))
student["Science"] = int(input("Science Marks: "))
student["Computer"] = int(input("Computer Marks: "))
student["Hindi"] = int(input("Hindi Marks: "))

total = (student["Math"] + student["English"] +
         student["Science"] + student["Computer"] +
         student["Hindi"])

percentage = (total / 500) * 100

def grade(mark):
    if mark >= 90:
        return "A+"
    elif mark >= 80:
        return "A"
    elif mark >= 70:
        return "B"
    elif mark >= 60:
        return "C"
    elif mark >= 40:
        return "D"
    else:
        return "Fail"

print("\n----- marksheet -----")
print("Math      :", student["Math"], " ", grade(student["Math"]))
print("English   :", student["English"], " ", grade(student["English"]))
print("Science   :", student["Science"], " ", grade(student["Science"]))
print("Computer  :", student["Computer"], " ", grade(student["Computer"]))
print("Hindi     :", student["Hindi"], " ", grade(student["Hindi"]))
print("Total     :", total ," / 500")
print("Percentage:", percentage, "%")
```
